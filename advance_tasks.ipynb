```python
# Task 9: Message Analyzer

message = input("Enter your message: ").title()

total_character = len(message)

total_words = len(message.split())

vowels = "aeiouAEIOU"
vowels_count = 0
for vowel in message:
  if vowel in vowels:
    vowels_count += 1

total_spaces = message.count(" ")

upper_case = message.upper()

if "python" in message.lower():
  python_exist = "yes"
else:
  python_exist = "no"

print("========== Message Analysis ==========")


print("Message: ", message)
print("Total character: ", total_character)
print("Total words: ", total_words)
print("Total vowel: ", vowels_count)
print("Total spaces: ", total_spaces)
print("Uppercase_formating: ", upper_case)
print("Python Exist: ", python_exist)
```

```python
#task 10

contacts = {}

while True:
    print("\n1. Add Contact")
    print("2. Search Contact")
    print("3. Display All Contacts")
    print("4. Exit")

    choice = input("Enter your choice: ")

    # Add Contact
    if choice == "1":
        name = input("Enter contact name: ")
        phone = input("Enter phone number: ")

        if name in contacts:
            print("Contact already exists!")
        else:
            contacts[name] = phone
            print("Contact added successfully!")

    # Search Contact
    elif choice == "2":
        name = input("Enter contact name to search: ")

        if name in contacts:
            print("Phone Number:", contacts[name])
        else:
            print("Contact not found!")

    # Display All Contacts
    elif choice == "3":
        if len(contacts) == 0:
            print("No contacts available!")
        else:
            print("\nAll Contacts:")
            for name, phone in contacts.items():
                print("Name:", name, "| Phone:", phone)

    # Exit
    elif choice == "4":
        print("Exiting Contact Book...")
        break

    else:
        print("Invalid choice! Please try again.")
```

```python
#task 11
#advance tasks


total_expense = int(input("How many expenses do you want to add: "))

expenses_list = []

for i in range(1, total_expense + 1):
  name = input("Enter your expense name: ")
  category = input("Enter your expense category: ")
  amount = float(input("Enter your expense amount: "))

  expense_item = {
      "expense_name" : name,
      "category" : category,
      "amount": amount
  }

  expenses_list.append(expense_item)

total_spending = 0
for expense_item in expenses_list:
  total_spending = total_spending + expense_item["amount"]

category_spending = {
    "Food": 0,
    "Transport": 0,
    "Shopping": 0,
    "Bills": 0,
    "Other": 0
}

for expense_item in expenses_list:
    category = expense_item["category"]
    amount = expense_item["amount"]

    if category in category_spending:
        category_spending[category] = category_spending[category] + amount
    else:
        category_spending["Other"] = category_spending["Other"] + amount

highest_expense = expenses_list[0]
for expense_item in expenses_list:

    if expense_item["amount"] > highest_expense["amount"]:
        highest_expense = expense_item

budget = float(input("Enter your budget: "))

if total_spending > budget:
    print("Warning! Budget exceeded.")
else:
    print("You are within your budget.")

def final_report():

    print("\n----- Expense Report -----")

    print("Total Spending:", total_spending)

    print("\nCategory Spending")

    for category, amount in category_spending.items():
        print(category, ":", amount)

    print("\nHighest Expense")
    print("Name:", highest_expense["expense_name"])
    print("Amount:", highest_expense["amount"])

final_report()
```

```python
# task 12


products = {}


# 1. Add Product
def add_product():
    name = input("Enter product name: ")

    if name in products:
        print("Product already exists!")
        return

    price = float(input("Enter product price: "))
    quantity = int(input("Enter available quantity: "))

    products[name] = {
        "price": price,
        "quantity": quantity
    }

    print("Product added successfully!")


# 2. Sell Product
def sell_product():
    name = input("Enter product name: ")

    if name not in products:
        print("Product not found!")
        return

    quantity = int(input("Enter quantity to sell: "))

    if quantity > products[name]["quantity"]:
        print("Insufficient stock!")
    else:
        products[name]["quantity"] -= quantity
        print("Product sold successfully!")


# 3. Restock Product
def restock_product():
    name = input("Enter product name: ")

    if name not in products:
        print("Product not found!")
        return

    quantity = int(input("Enter quantity to add: "))

    products[name]["quantity"] += quantity

    print("Product restocked successfully!")


# 4. Search Product
def search_product():
    name = input("Enter product name to search: ")

    if name in products:
        print("Product Name:", name)
        print("Price:", products[name]["price"])
        print("Quantity:", products[name]["quantity"])
    else:
        print("Product not found!")


# 5. Display Inventory
def display_inventory():
    if len(products) == 0:
        print("Inventory is empty!")
        return

    print("\n--- Inventory ---")

    total_value = 0

    for name, details in products.items():
        price = details["price"]
        quantity = details["quantity"]

        product_value = price * quantity
        total_value += product_value

        print("\nProduct:", name)
        print("Price:", price)
        print("Quantity:", quantity)
        print("Product Value:", product_value)

        # Low stock condition
        if quantity < 5:
            print("WARNING: Low Stock!")


    print("\nTotal Inventory Value:", total_value)


# Main Menu
while True:

    print("\n===== STORE INVENTORY MANAGER =====")
    print("1. Add Product")
    print("2. Sell Product")
    print("3. Restock Product")
    print("4. Search Product")
    print("5. Display Inventory")
    print("6. Exit")

    choice = input("Enter your choice: ")

    if choice == "1":
        add_product()

    elif choice == "2":
        sell_product()

    elif choice == "3":
        restock_product()

    elif choice == "4":
        search_product()

    elif choice == "5":
        display_inventory()

    elif choice == "6":
        print("Exiting Inventory Manager...")
        break

    else:
        print("Invalid choice! Please try again.")
```

```python
#task 13


books = []


# 1. Add Book
def add_book():
    book_id = input("Enter Book ID: ")

    # Check unique Book ID
    for book in books:
        if book["id"] == book_id:
            print("Book ID already exists!")
            return

    title = input("Enter Book Title: ")
    author = input("Enter Author Name: ")

    book = {
        "id": book_id,
        "title": title,
        "author": author,
        "available": True
    }

    books.append(book)

    print("Book added successfully!")


# 2. Display Books
def display_books():
    if len(books) == 0:
        print("No books available!")
        return

    print("\n--- All Books ---")

    for book in books:
        print("Book ID:", book["id"])
        print("Title:", book["title"])
        print("Author:", book["author"])

        if book["available"]:
            print("Status: Available")
        else:
            print("Status: Borrowed")

        print("------------------")


# 3. Search Book
def search_book():
    title = input("Enter book title to search: ")

    found = False

    for book in books:
        if book["title"].lower() == title.lower():
            print("\nBook Found!")
            print("Book ID:", book["id"])
            print("Title:", book["title"])
            print("Author:", book["author"])

            if book["available"]:
                print("Status: Available")
            else:
                print("Status: Borrowed")

            found = True
            break

    if not found:
        print("Book not found!")


# 4. Borrow Book
def borrow_book():
    book_id = input("Enter Book ID to borrow: ")

    for book in books:
        if book["id"] == book_id:

            if book["available"]:
                book["available"] = False
                print("Book borrowed successfully!")
            else:
                print("Book is already borrowed!")

            return

    print("Book not found!")


# 5. Return Book
def return_book():
    book_id = input("Enter Book ID to return: ")

    for book in books:
        if book["id"] == book_id:

            if book["available"]:
                print("This book is already available!")
            else:
                book["available"] = True
                print("Book returned successfully!")

            return

    print("Book not found!")


# Main Menu
while True:

    print("\n===== LIBRARY BOOK MANAGEMENT SYSTEM =====")
    print("1. Add Book")
    print("2. Display Books")
    print("3. Search Book")
    print("4. Borrow Book")
    print("5. Return Book")
    print("6. Exit")

    choice = input("Enter your choice: ")

    if choice == "1":
        add_book()

    elif choice == "2":
        display_books()

    elif choice == "3":
        search_book()

    elif choice == "4":
        borrow_book()

    elif choice == "5":
        return_book()

    elif choice == "6":
        print("Exiting Library System...")
        break

    else:
        print("Invalid choice! Please try again.")
```
